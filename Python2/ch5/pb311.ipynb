<!-- Diabetes Prediction using Random Forest
You are given a dataset diabetes.csv containing medical records of patients.
The dataset includes predictor variables such as BMI, Age, Insulin, Pregnancies, etc., and a target variable named Outcome.
Your task is to:
Load the dataset using pandas.
Separate independent and dependent variables.
Split dataset into training and testing sets.
Train a Random Forest Classifier.
Predict patient outcomes.
Evaluate performance using:
Confusion Matrix
Accuracy Score -->

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error,accuracy_score,confusion_matrix
from sklearn.ensemble import RandomForestClassifier
```

```python
df=pd.read_csv("Diabetes.csv")
df
```

```python
df.isna().sum()
```

```python
df.columns
```

```python
x=df[['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin',
       'BMI', 'DiabetesPedigreeFunction', 'Age']]
y=df['Outcome']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
x_test.shape
```

```python
nn=RandomForestClassifier(n_estimators=100,random_state=42,max_depth=5)
```

```python
model=nn.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
print(confusion_matrix(y_test,y_pred))
```
