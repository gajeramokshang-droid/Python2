```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("StudentsPerformance.csv")
```

```python
math_pass=[]
```

```python
for score in df['math score']:
    if score >= 50:
        math_pass.append(1)
    else:
        math_pass.append(0)
df['Math_Pass'] = math_pass
```

```python
reading_pass = []
for score in df['reading score']:
    if score >= 50:
        reading_pass.append(1)
    else:
        reading_pass.append(0)
df['Reading_Pass'] = reading_pass
```

```python
writing_pass = []
for score in df['writing score']:
    if score >= 50:
        writing_pass.append(1)
    else:
        writing_pass.append(0)
df['Writing_Pass'] = writing_pass
```

```python
df['Overall_Pass'] = df.apply(lambda row: 1 
                              if (row['Math_Pass'] == 1 and 
                                  row['Reading_Pass'] == 1 and 
                                  row['Writing_Pass'] == 1) else 0, axis=1)
```

```python
df.drop(['math score', 'reading score', 'writing score'], axis=1, inplace=True)
```

```python
category=['gender', 'race/ethnicity', 'parental level of education', 'lunch', 'test preparation course']
```

```python
df=pd.get_dummies(df, columns=category,drop_first=True)
```

```python
df.isna().sum()
```

```python
x=df.drop(['Math_Pass'], axis=1) 
y=df['Math_Pass']
```

```python
x_train,x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=10)
```
