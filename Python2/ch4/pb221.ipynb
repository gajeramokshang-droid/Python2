```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
df=pd.read_csv("data.csv")
df
```

```python
a=LinearRegression()
x=df[["Duration"]]
y=df['Pulse']
a.fit(x,y)
print("X:",a.coef_)
print("C:",a.intercept_)
y_predict=a.predict(x)
print(y_predict)
a.predict([[50]])
# - During training, you gave it x = df[["Duration"]] → notice that’s a 2D array (rows × columns).
# - So when predicting  a.predict([[]]), you must also give a 2D array, even if it’s just one value.
```
