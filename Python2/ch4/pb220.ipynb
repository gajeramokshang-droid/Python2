```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
df=pd.read_csv("placement.csv")
df
```

```python
x=df[['cgpa']] #independent
y=df['placement_exam_marks']  # Depenedent
```

```python
a=LinearRegression()
a.fit(x,y)
print("x:",a.coef_)
print("y",a.intercept_)
```
