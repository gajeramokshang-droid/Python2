```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
```

## <font color='red'># step3:  Data Preprocessing</font>

```python
df['age']=2026-df['Year']
df.head()
```

```python
df.drop(columns='Year',inplace=True)
df
```

```python
df
```

```python
df=pd.get_dummies(data=df,drop_first=True)
df.head()
```

```python
x=df[["Present_Price","Kms_Driven","Owner","age","Fuel_Type_Diesel","Fuel_Type_Petrol","Seller_Type_Individual","Transmission_Manual"]]
y=df['Selling_Price']
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.1, random_state=4)
a=LinearRegression()
a.fit(x_train,y_train)
print("Intercept",a.intercept_)
print("Coeff",a.coef_)
y_predi=a.predict(x_test)
mse=mean_squared_error(y_predi,y_test)
print("MSE",mse)

print("MAE=",mean_absolute_error(y_test,y_predi))
```

```python

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
df = pd.read_csv('RealEstate.csv')
df
```

```python

df = df.drop(columns=['No'])
print("\nNull values in features:")
print(df.isnull().sum())

X = df[['X1 transaction date', 'X2 house age', 'X3 distance to the nearest MRT station',
        'X4 number of convenience stores', 'X5 latitude', 'X6 longitude']]
y = df['Y house price of unit area']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=110)

model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print("\nPredictions for the testing set:")
print(y_pred)

print("\nCoefficients of the regression model:")
print(model.coef_)
print("\nMean squared error of the regression model:")
print(mean_squared_error(y_test, y_pred))
```


## <font color='red'>polynomial Linear Regression</font>

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import PolynomialFeatures
```

```python
df=pd.read_csv("Ice_cream selling data.csv")
df
```

```python
x=df[['Temperature (°C)']]   #INDEPENDENT
y=df['Ice Cream Sales (units)']   #DEPENDENT
```

```python
#Split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
poly=PolynomialFeatures(degree=3)
x_poly=poly.fit_transform(x_train)
```

```python
a=LinearRegression()
```

```python
a.fit(x_poly,y_train)
```

```python
x_test_poly = poly.transform(x_test)  # Transform test set
y_pred=a.predict(x_test_poly)
print("M:",a.coef_)
print("C:",a.intercept_)
```

```python
print("MSE",mean_squared_error(y_test,y_pred))
```
