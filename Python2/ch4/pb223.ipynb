```python
import pandas as pd
df=pd.read_csv("auto.csv")
df
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
```

```python
X=df[["acceleration"]]
y=df['model year']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=40)
```

```python
a=LinearRegression()


# Raw Data (X_train, y_train)
#         ↓
# OLS Algorithm runs
#         ↓
# Finds slope(s) (m) and intercept (c)
#         ↓
# Stores them in a.coef_ and a.intercept_
#         ↓
# Model is now trained → ready for predictions


a.fit(X_train,y_train)  # IT FOUND M AND C
print("Slope(m)",a.coef_)
print("INtercept(c)",a.intercept_)
```

```python
y_pred=a.predict(X_test)
y_pred
```

```python
mse=mean_squared_error(y_test,y_pred)
mse
```

```python
print(a.score(X_test,y_test))
```
