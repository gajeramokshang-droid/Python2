```python
# Predict salary based on job position of 6.5 using polynomial regression with a degree of 3 and scikit learn library for the given
# 'Position_Salaries.csv' dataset. (Don't split data for training/testing)
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

# Load the dataset
df = pd.read_csv('Position_Salaries.csv')
df
# Extract features (X) and target (y)
X = df[['Position']].values  # Ensure X is a 2D array
y = df['Salary'].values

# Create polynomial features (degree=3)
poly = PolynomialFeatures(degree=3)
X_poly = poly.fit_transform(X)

# Fit linear regression on polynomial features
model = LinearRegression()
model.fit(X_poly, y)

# Predict salary for position 6.5
position_6_5 = poly.transform([[6.5]])  # Transform to polynomial features
predicted_salary = model.predict(position_6_5)

print(f"Predicted Salary for Position 6.5: {predicted_salary[0]:.2f}")
```
