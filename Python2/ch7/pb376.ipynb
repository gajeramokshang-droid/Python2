```python
# Write a program of web scrapping using BeautifulSoup to scrape the given data from the following link.
# https://editorial.rottentomatoes.com/guide/popular-movies/
# On the above link, you’ll find 30 Popular movies. Scrape the Movie Title and Rating of that particular movie and make a
# Dataframe of the same
```

```python
import requests
from bs4 import BeautifulSoup
website='https://editorial.rottentomatoes.com/guide/popular-movies/'
response=requests.get(website)
content=response.text
```

```python
soup=BeautifulSoup(content,'html.parser')
christ=soup.find_all('div',class_='layout')[:30]
print(christ)
```

```python
michael = []
for item in christ:
    statement_div = item.find('div', class_='meta-title-wrapper')
    mik = statement_div.find('a', class_='meta-title') if statement_div else None
    rating_div = item.find('div', class_='meta-scores-wrapper')
    mary = rating_div.find('span', class_='tMeterScore') if rating_div else None

    if mik and mary:
        statement = mik.get_text(strip=True)
        ratings = mary.get_text(strip=True)
        # Use a dictionary with keys
        michael.append({"Title": statement, "Rating": ratings})
```

```python
import pandas as pd
df=pd.DataFrame(michael)
print(df)
```
