```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
# carname,mileage,price,waiting
```

```python
with open('Certified used Mercedes-Benz for sale.html','r',encoding='utf-8') as file:
    content=file.read()
soup=BeautifulSoup(content,'html.parser')
```

```python
cont=soup.find_all('div',class_='vehicle-details')
samson=[]
for i in cont:
    con=i.find('h2',class_='title').text.strip()
    con1=i.find('div',class_='mileage').text.strip()
    con2=i.find('div',class_='price-section price-section-vehicle-card').text.strip()
    con4=i.find('span',class_='sds-rating__count')
    samson.append((con,con1,con2,con4))    
```

```python
df=pd.DataFrame(samson)
df                                                                                      
```
