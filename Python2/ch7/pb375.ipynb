```python
# Write a program for web scrapping using BeautifulSoup to scrape the following details from the given link and make a data
# frame using that scraped data from
# the page in a given link.
# Link: https://www.politifact.com/factchecks
# You will find 30 news articles with fact checks on this page. You need to scrape the following details from all the articles and
# store that in a data frame. Statement of News, Date of News, Source of News
```

```python
from bs4 import BeautifulSoup
import pandas as pd
import requests

# URL of the Politifact page
url = 'https://www.politifact.com/factchecks'
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

# Find the list items containing the fact-checks
articles = soup.find_all('li', class_='o-listicle__item')[:30]

scraped_data = []

for article in articles:
    # Extract Statement
    statement_div = article.find('div', class_='m-statement__storyandtext')
    statement = statement_div.find('a').text.strip() if statement_div and statement_div.find('a') else "N/A"
    
    # Extract Source
    source_element = article.find('a', class_='m-statement__author-name')
    source = source_element.text.strip() if source_element else "N/A"
    
    # Extract Date
    footer_element = article.find('footer', class_='m-statement__footer')
    date = footer_element.text.split('•')[-1].strip() if footer_element else "N/A"
    
    scraped_data.append({
        "Statement of News": statement,
        "Source of News": source,
        "Date of News": date
    })

# Create DataFrame
df = pd.DataFrame(scraped_data)
print(df)

```
