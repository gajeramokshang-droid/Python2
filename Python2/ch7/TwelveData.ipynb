```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url='https://twelvedata.com/'
response=requests.get(url)
soup=BeautifulSoup(response.text,'html.parser')
jesus=soup.find_all("div",class_='rates-line-prices rates-line-prices--gray')
stock_market=[]
for i in jesus:
    price = i.find_all('span', class_='symbol-price symbol-price-js')
    symbol = i.find_all('span', class_='symbol-changes-js')
    for span in price:
        price=span.text.strip()

    for span in symbol:
        symbol=span.text.strip()
    for prices,symbols in zip(price,symbol):
        stock_market.append({
            "price":prices,
            'Symbol':symbol
        })
platform=soup.find_all('div',class_='col col-lg-7 col-12')

information=[]
for k in platform:
    singleplatform=k.find('span',class_='accent').text.strip()
    Topic=k.find('h3').text.strip()
    data=k.find_all('p')
    links=[]
    for i in data:
        mokshang=i.find_all('a')
        for a in mokshang:
            links.append(a.text.strip())
    information.append({
        'singleplatform':singleplatform,
        'topic':Topic,
        "Data":links
    })

print(information)
print(stock_market)

df=pd.DataFrame(information,stock_market)
df
```
