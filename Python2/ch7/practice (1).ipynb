```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
with open('30 Most Popular Movies.html','r',encoding='utf-8') as file:
    content=file.read()
soup=BeautifulSoup(content,'html.parser')
```

```python
cont=soup.find_all('div',class_='article_movie_title')[:30]
movue=[]
for i in cont:
    con=i.find('a').text.strip()
    con1=i.find('span',class_='tMeterScore').text.strip()
    movue.append((con,con1))
print(movue)
```

```python
df=pd.DataFrame(movue)
print(df)
```
