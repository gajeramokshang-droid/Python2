```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
with open('flipkart.html','r',encoding='utf-8') as file:
    content=file.read()
soup=BeautifulSoup(content,'html.parser')
```

```python
name=soup.find_all('a',class_='_1fQZEK')
jk=[]
for i in name:
    mok=i.find('div',class_='_4rR01T').text.strip()
    gaj=i.find('div',class_='_3LWZlK').text.strip()
    discount=i.find('div',class_='_30jeq3 _1_WHN1')
    actual=i.find('div',class_='_3I9_wc _27UcVY')
    if not discount or not actual:
        continue
    dis=int(discount.text.strip().replace(',', '').replace('₹',"")) 
    actu=int(actual.text.strip().replace(',', '').replace('₹', ''))
    difference=actu-dis
    jk.append((mok,gaj,discount,actu,int(difference)))
```

```python
df=pd.DataFrame(jk)
df
```
