```python
import requests 
from bs4 import BeautifulSoup
```

```python
website="https://subslikescript.com/movies"
result=requests.get(website)
result  #object format
```

```python
content=result.text
soup=BeautifulSoup(content,'html.parser')
```

```python
box=soup.find("article",class_='main-article')
box
```

```python
movie_title=box.find_all("a")  #return in list
movie_title

list=[]
for i in movie_title:
    titl=i.text
    list.append(titl)
list
```

```python
links=[]
for i in box.find_all('a'):
    links.append(i['href'])
links
```

```python
route = "https://subslikescript.com"
for i in links:
    website = route + i
    result = requests.get(website)
    content = result.text
    soup = BeautifulSoup(content, 'html.parser')
    transcript = soup.find('div', class_='full-script').text
    T = soup.find('h1').text.strip()
    with open(T + '.txt', 'w', encoding='utf-8') as file:
        file.write(transcript)
```

```python
import requests 
from bs4 import BeautifulSoup
import pandas as pd
```

```python
website="https://webscraper.io/test-sites/e-commerce/allinone/computers/laptops"
result=requests.get(website)
result  #object format
```

```python
content=result.text
soup=BeautifulSoup(content,'html.parser')
```

```python
laptopelement = soup.find_all("a", class_='title')
priceelement=soup.find_all("span")

laptopss=[]
pricess=[]
for laptops,prices in zip(laptopelement,priceelement):
    laptopss.append(laptops.text.strip())
    pricess.append(prices.text.strip())

df = pd.DataFrame({
    'Laptop': laptopss,
    'Price': pricess
})
df
```
