```python
import requests 
from bs4 import BeautifulSoup
```

```python
website='https://subslikescript.com/movie/Atonement-783233'
result=requests.get(website)
result  #object format
```

```python
content=result.text
```

```python
soup=BeautifulSoup(content,'html.parser')
```

```python
title=soup.find('h1')
title.text
```

```python
title=soup.find('p')
title.text
```

```python
title=soup.find('div')
title.text
```

```python
transcript=soup.find('div',class_='full-script').text
transcript
```

```python
with open('Atonement.txt','w',encoding='utf-8') as file:
    file.write(transcript)
```
