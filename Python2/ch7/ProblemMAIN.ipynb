```python
import requests
from bs4 import BeautifulSoup
```

```python
website='https://www.foxnews.com/'
result=requests.get(website)
```

```python
result
```

```python
content=result.text
```

```python
soup=BeautifulSoup(content,'html.parser')
soup.text
```

```python
heading=soup.find_all(['h1','h2','h3','h4','h5','h6'])
for headings in heading:
    print(headings.text.strip())
```

```python
politics=soup.find_all('div',class_='page-content')
for politic in politics:
    print(politic.text.strip())
```
