```python
import requests
from bs4 import BeautifulSoup

website='https://www.biblegateway.com/'
result=requests.get(website)
print(result)

content=result.text
print("Content:",content)

soup=BeautifulSoup(content,'html.parser')
print(soup)
```

```python
title=soup.find('div',class_='mast item')
title.text
```

```python
import requests
from bs4 import BeautifulSoup
```

```python
website='https://www.foxnews.com/'
result=requests.get(website)
result
```

```python
content=result.text
```

```python
soup=BeautifulSoup(content,'html.parser')
```

```python
find=soup.find('h2',class_='title title-color-red')
find.text
```

```python
heading=soup.find_all(['h1','h2','h3','h4','h5','h6'])
for headings in heading:
    print(headings.text.strip())
```

```python
find=soup.find('div',class_='m')
find.text
```

```python
politics=soup.find('div',class_='content')
for politic in politics:
    print(politic.text.strip())
```
