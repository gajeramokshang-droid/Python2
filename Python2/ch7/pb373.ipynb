```python
import requests
import pandas as pd

api_key = "76bbb98de15cb7d16f0a7b4644071b52"
lat = 23.0225
lon = 72.5714

# Forecast API (5 days, 3-hour intervals)
url = f"https://api.openweathermap.org/data/2.5/forecast?lat={lat}&lon={lon}&appid={api_key}&units=metric"

response = requests.get(url)
data = response.json()

# Extract forecast entries
forecast_list = data["list"]

# Build rows for DataFrame
rows = []
for entry in forecast_list:
    rows.append({
        "date_time": entry["dt_txt"],
        "min_temp": entry["main"]["temp_min"],
        "max_temp": entry["main"]["temp_max"],
        "wind_speed": entry["wind"]["speed"],
        "humidity": entry["main"]["humidity"],
        "weather_description": entry["weather"][0]["description"]
    })

# Create DataFrame
df = pd.DataFrame(rows, columns=["date_time", "min_temp", "max_temp", "wind_speed", "humidity", "weather_description"])

# Display first 10 rows
print(df.head(10))
```
